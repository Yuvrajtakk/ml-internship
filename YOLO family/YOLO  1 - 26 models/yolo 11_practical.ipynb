# YOLO11 Practical Notebook — 2024

Loads a real pretrained YOLO11 model and runs it on a fixed test image. Practical/comparison-focused, not architecture theory (see `YOLO_family.ipynb` for that).

### What's new vs. the previous version

- Introduces the C3k2 block (more efficient backbone unit than v8's C2f)
- Adds C2PSA: partial self-attention, letting the model weigh important regions without full transformer cost
- Iterative refinement release, not a paradigm shift

### Step 3: Load the pretrained model

````python
from ultralytics import YOLO

model = YOLO('yolo11n.pt')
print('Loaded:', 'yolo11n.pt')
print('Classes:', len(model.names))
````

### Step 4: Run inference + time it

````python
import time

start = time.time()
results = model('bus.jpg')
elapsed = time.time() - start

print(f"Inference time: {elapsed:.3f}s")
print(f"Objects detected: {len(results[0].boxes)}")
for box in results[0].boxes:
    cls_name = model.names[int(box.cls[0])]
    conf = float(box.conf[0])
    print(f"  {cls_name}: {conf:.2f}")

results[0].show()
````
