This is a comparison-only notebook using verified published benchmarks, plus one illustrative (not real-model) visualization.

### What's new vs. "the old way" (pre-YOLO detectors)

- Single-pass grid regression: one forward pass predicts all boxes at once, instead of running a classifier separately on ~2000 region proposals
- Real-time capable for the first time: 45 FPS (base model), 155 FPS (Fast YOLO variant, 9 conv layers instead of 24)
- Tradeoff: only 49 grid cells x 2 boxes = hard cap of 49 objects/image, and shared class-per-cell (can't detect 2 different-class objects whose centers fall in the same cell)

### Verified benchmarks (PASCAL VOC 2007+2012)

| Model | mAP | Speed |
|---|---|---|
| DPM (fastest variant) | 30.4% | 15 FPS |
| YOLOv1 (base) | 63.4% | 45 FPS |
| Fast YOLO | 52.7% | 155 FPS |
| Faster R-CNN (VGG-16) | 73.2% | 7 FPS |

YOLOv1 broke the speed/accuracy tradeoff that defined the field until 2015 -- not the most accurate, but dramatically faster than anything close to its accuracy.

### Illustrative visualization (NOT a real trained model)

This shows the grid concept on the shared test image -- an illustrative overlay, not genuine detections.

```python
import cv2
import matplotlib.pyplot as plt

!wget -q https://ultralytics.com/images/bus.jpg -O bus.jpg

img = cv2.cvtColor(cv2.imread('bus.jpg'), cv2.COLOR_BGR2RGB)
h, w = img.shape[:2]

# Overlay a 7x7 grid (YOLOv1's real grid size) just to show the CONCEPT --
# this draws no real detections, since there's no real model here.
plt.figure(figsize=(8, 8))
plt.imshow(img)
for i in range(8):
    plt.axhline(i * h / 7, color='yellow', linewidth=1, alpha=0.6)
    plt.axvline(i * w / 7, color='yellow', linewidth=1, alpha=0.6)
plt.title("YOLOv1's 7x7 grid concept (illustrative overlay, not real detections)")
plt.axis('off')
plt.show()
```

### One-line takeaway

YOLOv1 proved real-time object detection was possible by reframing it as one-shot grid regression -- but its 49-cell hard cap and shared per-cell class prediction were real, measurable limitations later versions fixed one at a time.
