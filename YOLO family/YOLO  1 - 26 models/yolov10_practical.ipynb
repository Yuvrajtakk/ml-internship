# YOLOv10 Practical Notebook — 2024

Loads a real pretrained YOLOv10 model and runs it on a fixed test image. Practical/comparison-focused, not architecture theory (see `YOLO_family.ipynb` for that).

### What's new vs. the previous version

- Trains TWO heads jointly: one-to-many (rich training signal) and one-to-one (deployed)
- The one-to-one head is trained to never produce duplicate detections
- Result: NMS (Non-Max Suppression) is no longer needed at inference -- lower latency

### Step 3: Load the pretrained model

````python
from ultralytics import YOLO

model = YOLO('yolov10n.pt')
print('Loaded:', 'yolov10n.pt')
print('Classes:', len(model.names))
````

### Step 4: Run inference + time it

````python
import time

start = time.time()
results = model('bus.jpg')
elapsed = time.time() - start

print(f"Inference time: {elapsed:.3f}s")
print(f"Objects detected: {len(results[0].boxes)}")
for box in results[0].boxes:
    cls_name = model.names[int(box.cls[0])]
    conf = float(box.conf[0])
    print(f"  {cls_name}: {conf:.2f}")

results[0].show()
````
