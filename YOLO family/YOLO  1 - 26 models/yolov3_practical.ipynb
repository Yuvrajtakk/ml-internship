# YOLOv3 Practical Notebook — 2018

Loads a real pretrained YOLOv3 model and runs it on a fixed test image. Practical/comparison-focused, not architecture theory (see `YOLO_family.ipynb` for that).

### What's new vs. the previous version

- Darknet-53 backbone (residual connections, much deeper than v2's Darknet-19)
- Multi-scale prediction: detects at 3 grid resolutions (13x13, 26x26, 52x52) instead of 1
- Switched from softmax to independent logistic classifiers -- supports overlapping/multi-label classes

### Step 3: Load the pretrained model

````python
from ultralytics import YOLO

model = YOLO('yolov3u.pt')
print('Loaded:', 'yolov3u.pt')
print('Classes:', len(model.names))
````

### Step 4: Run inference + time it

````python
import time

start = time.time()
results = model('bus.jpg')
elapsed = time.time() - start

print(f"Inference time: {elapsed:.3f}s")
print(f"Objects detected: {len(results[0].boxes)}")
for box in results[0].boxes:
    cls_name = model.names[int(box.cls[0])]
    conf = float(box.conf[0])
    print(f"  {cls_name}: {conf:.2f}")

results[0].show()
````
