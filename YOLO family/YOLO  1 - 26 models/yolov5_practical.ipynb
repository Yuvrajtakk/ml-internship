### What's new vs. the previous version

- PyTorch reimplementation (v1-v4 were Darknet/C)
- AutoAnchor: automatically recomputes anchors per-dataset
- Released in n/s/m/l/x sizes for speed/accuracy tradeoff
- Fixed the grid-sensitivity math bug from v2-v4's sigmoid formula

### Step 3: Load the pretrained model

````python
from ultralytics import YOLO

model = YOLO('yolov5nu.pt')
print('Loaded:', 'yolov5nu.pt')
print('Classes:', len(model.names))
````

### Step 4: Run inference + time it

````python
import time

start = time.time()
results = model('bus.jpg')
elapsed = time.time() - start

print(f"Inference time: {elapsed:.3f}s")
print(f"Objects detected: {len(results[0].boxes)}")
for box in results[0].boxes:
    cls_name = model.names[int(box.cls[0])]
    conf = float(box.conf[0])
    print(f"  {cls_name}: {conf:.2f}")

results[0].show()
````
