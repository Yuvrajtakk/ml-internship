v6 isn't part of the official YOLO lineage (different authors, not merged into `ultralytics`), and its repo runs inference via a CLI script rather than a clean Python class. Practical/comparison-focused, not architecture theory.

### What's new vs. the previous version

- Anchor-free detection (ahead of v8 doing the same thing)
- Decoupled head: separate branches for classification and box regression
- RepVGG-style reparameterization: trains with a richer multi-branch structure, mathematically collapses to one fast branch at inference -- training-time accuracy benefit with zero inference-time cost

### Step 3 & 4: Run inference via the repo's own script + time it

Unlike the `ultralytics`-based versions, v6 doesn't expose a simple `YOLO()` Python class -- inference runs through their own `infer.py` script, writing an annotated image to a `runs/` folder.

````python
import time

start = time.time()
!python YOLOv6/tools/infer.py --weights yolov6s.pt --source bus.jpg --save-txt --project runs --name v6_result --exist-ok
elapsed = time.time() - start
print(f"Approx inference (incl. script startup) time: {elapsed:.3f}s")
````

### Step 5: Display the annotated result

````python
from IPython.display import Image, display
import glob

result_images = glob.glob('runs/v6_result/*.jpg')
if result_images:
    display(Image(filename=result_images[0]))
else:
    print("No output image found -- check the infer.py command output above for errors.")
````
